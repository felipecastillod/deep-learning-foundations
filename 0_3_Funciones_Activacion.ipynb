# Funciones de Activación en Deep Learning

## Module 1 — Neural Foundations

Este notebook estudia las funciones de activación más importantes en Deep Learning: **Sigmoid, Tanh, ReLU y Leaky ReLU**.

Además, se analizan conceptos fundamentales como:

- No linealidad
- Vanishing Gradient
- Dead ReLU
- Redes neuronales profundas
- Aplicaciones modernas en Deep Learning


# ¿Por qué necesitamos funciones de activación?

Una neurona artificial realiza inicialmente una operación lineal:

y = W·X + b

Si una red neuronal estuviera compuesta únicamente por operaciones lineales, múltiples capas seguirían siendo equivalentes a una sola transformación lineal.

Las funciones de activación introducen no linealidad y permiten aprender patrones complejos.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10,10,1000)
```

# Función Sigmoid

La función Sigmoid transforma cualquier valor real al intervalo (0,1).

Aplicaciones históricas:
- Regresión logística
- Redes neuronales tempranas
- Interpretación probabilística


```python
sigmoid = 1/(1+np.exp(-x))

plt.figure(figsize=(8,5))
plt.plot(x,sigmoid)
plt.title("Función Sigmoid")
plt.grid(True)
plt.show()
```

## Problema: Vanishing Gradient

Cuando los valores son muy grandes o muy pequeños, la pendiente de Sigmoid se aproxima a cero.

Esto dificulta el aprendizaje en redes profundas.


# Función Tanh

Tanh mejora algunas limitaciones de Sigmoid.

Ventajas:
- Salida entre -1 y 1
- Centrada en cero
- Mejor flujo de gradientes


```python
tanh = np.tanh(x)

plt.figure(figsize=(8,5))
plt.plot(x,tanh)
plt.title("Función Tanh")
plt.grid(True)
plt.show()
```

# Función ReLU

ReLU es actualmente una de las funciones más utilizadas.

Definición:

f(x)=max(0,x)

Ventajas:
- Simple
- Rápida
- Reduce Vanishing Gradient


```python
relu = np.maximum(0,x)

plt.figure(figsize=(8,5))
plt.plot(x,relu)
plt.title("Función ReLU")
plt.grid(True)
plt.show()
```

## Problema: Dead ReLU

Cuando una neurona permanece mucho tiempo en la región negativa:

- Salida = 0
- Gradiente = 0

La neurona puede dejar de aprender.


# Función Leaky ReLU

Leaky ReLU introduce una pequeña pendiente para valores negativos.

Esto evita que el gradiente desaparezca completamente.


```python
leaky_relu = np.where(x>0,x,0.01*x)

plt.figure(figsize=(8,5))
plt.plot(x,leaky_relu)
plt.title("Función Leaky ReLU")
plt.grid(True)
plt.show()
```

# Comparación de todas las funciones


```python
plt.figure(figsize=(12,6))

plt.plot(x,sigmoid,label='Sigmoid')
plt.plot(x,tanh,label='Tanh')
plt.plot(x,relu,label='ReLU')
plt.plot(x,leaky_relu,label='Leaky ReLU')

plt.title("Comparación de Funciones de Activación")
plt.legend()
plt.grid(True)
plt.show()
```

# Comparación Conceptual

| Función | Rango | Ventaja Principal | Limitación |
|----------|--------|------------------|-------------|
| Sigmoid | 0 a 1 | Interpretación probabilística | Vanishing Gradient |
| Tanh | -1 a 1 | Centrada en cero | Saturación |
| ReLU | 0 a +∞ | Entrenamiento eficiente | Dead ReLU |
| Leaky ReLU | -∞ a +∞ | Evita neuronas muertas | Requiere parámetro α |
